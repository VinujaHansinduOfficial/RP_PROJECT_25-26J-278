```python
import os
import pandas as pd

FILENAME = "final_cleaned_overstimulation_dataset_90to95_values_modified.csv"

# Search for file automatically
def find_file(filename):
    for root, dirs, files in os.walk(".", topdown=True):
        if filename in files:
            return os.path.join(root, filename)
    return None

filepath = find_file(FILENAME)

if filepath:
    print(f" File found at: {filepath}")
    data = pd.read_csv(filepath)
    print(" Dataset Loaded Successfully!\n")
    print(f"Shape of dataset: {data.shape}")
else:
    print(" File NOT found. Check the file name or location.")
```

```python
# ============================================
# 🧹 STEP 2: Data Cleaning & Preprocessing
# ============================================
import numpy as np

# Make a copy to keep the original safe
df = data.copy()

#  Handle missing or null values
print("Missing values before cleaning:\n", df.isnull().sum())

# Impute numeric columns with median
num_cols = df.select_dtypes(include=[np.number]).columns
for col in num_cols:
    if df[col].isnull().any():
        df[col].fillna(df[col].median(), inplace=True)

# Impute categorical columns (if any)
cat_cols = df.select_dtypes(include=['object', 'category']).columns
for col in cat_cols:
    if df[col].isnull().any():
        df[col].fillna(df[col].mode()[0], inplace=True)

print("\n Missing values handled.\n")
```

```python

#  Remove duplicates
before = df.shape[0]
df.drop_duplicates(inplace=True)
after = df.shape[0]
print(f"Removed {before - after} duplicate rows.\n")
```

```python
#  Check and correct inconsistent / invalid entries
# Example: ensure values are within logical ranges
def fix_outliers(col, low=None, high=None):
    if col in df.columns:
        before = df[col].isnull().sum()
        df.loc[(df[col] < low) | (df[col] > high), col] = np.nan
        after = df[col].isnull().sum()
        if after > before:
            print(f"Column '{col}': replaced {after-before} out-of-range values with NaN.")
            df[col].fillna(df[col].median(), inplace=True)

fix_outliers('Age', 6, 120)
fix_outliers('Sleep_Hours', 0, 24)
fix_outliers('Screen_Time', 0, 24)
fix_outliers('Heart_Rate', 30, 220)

print("\n Outliers checked and corrected.\n")
```

```python
#  Convert data types where necessary
for col in df.columns:
    if df[col].dtype == 'object':
        try:
            df[col] = pd.to_numeric(df[col], errors='ignore')
        except:
            pass

print(" Data types verified/converted where possible.\n")
```

```python
# ============================================
#  STEP 4: Statistical Analysis (t-tests / Chi-square)
# ============================================
# Paste & run this cell after Step 2/3 (requires cleaned DataFrame 'df').

import pandas as pd
import numpy as np
from pathlib import Path
from scipy.stats import ttest_ind, levene, mannwhitneyu, chi2_contingency
import math
import matplotlib.pyplot as plt

# 0) Load df if kernel restarted
if 'df' not in globals():
    cleaned_path = Path("final_cleaned_overstimulation_dataset_90to95_values_modified.csv")
    if cleaned_path.exists():
        df = pd.read_csv(cleaned_path)
        print("Loaded cleaned dataset from", cleaned_path)
    else:
        raise FileNotFoundError("DataFrame 'df' not found. Run Step 2 first and ensure 'df' exists in the notebook.")

# 1) Detect target column
target_candidates = [c for c in df.columns if 'overstimulated' in c.lower()]
if not target_candidates:
    for cand in ('overstimulated','over_stimulated','over_stimulation','overstim'):
        if cand in [c.lower() for c in df.columns]:
            target_candidates = [c for c in df.columns if c.lower() == cand]
            break
if not target_candidates:
    raise KeyError("No target column matching 'Overstimulated' found. Columns:\n" + ", ".join(df.columns[:100]))
target = target_candidates[0]
print("Using target column:", target)

# 2) Ensure target is binary numeric (map if necessary)
if not set(df[target].dropna().unique()).issubset({0,1}):
    uniq = list(df[target].dropna().unique())
    if len(uniq) == 2:
        mapping = {uniq[0]:0, uniq[1]:1}
        df[target] = df[target].map(mapping)
        print("Mapped target to 0/1 using:", mapping)
    else:
        raise ValueError("Target column is not binary and not mappable automatically. Unique values:\n" + str(uniq))

# 3) Prepare groups
group0 = df[df[target] == 0]
group1 = df[df[target] == 1]

# 4) Feature lists
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
if target in num_cols:
    num_cols.remove(target)
cat_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()

# helper: Cohen's d
def cohens_d(x, y):
    x = np.asarray(x.dropna())
    y = np.asarray(y.dropna())
    nx, ny = len(x), len(y)
    if nx < 2 or ny < 2:
        return np.nan
    dof = nx + ny - 2
    pooled_var = ((nx - 1) * np.nanvar(x, ddof=1) + (ny - 1) * np.nanvar(y, ddof=1)) / dof
    if pooled_var <= 0:
        return np.nan
    return (np.nanmean(x) - np.nanmean(y)) / np.sqrt(pooled_var)

results = []

# 5) Numeric feature tests: Levene (variance), t-test (Welch if unequal), Mann-Whitney, Cohen's d
print("\nRunning tests for numeric features...")
for col in num_cols:
    try:
        if df[col].nunique(dropna=True) <= 1:
            results.append((col, 'numeric', np.nan, np.nan, np.nan, np.nan, 'constant_or_single_value'))
            continue
        x0 = group0[col].dropna()
        x1 = group1[col].dropna()
        if len(x0) < 3 or len(x1) < 3:
            results.append((col, 'numeric', np.nan, np.nan, np.nan, np.nan, 'insufficient_samples'))
            continue
        lev_stat, lev_p = levene(x0, x1)
        equal_var = lev_p >= 0.05
        t_stat, t_p = ttest_ind(x1, x0, equal_var=equal_var, nan_policy='omit')
        try:
            mw_stat, mw_p = mannwhitneyu(x1, x0, alternative='two-sided')
        except Exception:
            mw_stat, mw_p = (np.nan, np.nan)
        d = cohens_d(x1, x0)
        notes = f"levene_p={lev_p:.4f}"
        results.append((col, 'numeric', float(t_stat), float(t_p), float(mw_p) if not np.isnan(mw_p) else np.nan, float(d) if not np.isnan(d) else np.nan, notes))
    except Exception as e:
        results.append((col, 'numeric', np.nan, np.nan, np.nan, np.nan, f'error:{e}'))

# 6) Categorical features: Chi-square tests
print("\nRunning Chi-square for categorical features...")
for col in cat_cols:
    try:
        ct = pd.crosstab(df[col], df[target])
        if ct.shape[0] < 2 or ct.shape[1] < 2:
            results.append((col, 'categorical', np.nan, np.nan, np.nan, np.nan, 'insufficient_levels'))
            continue
        chi2, p, dof, expected = chi2_contingency(ct)
        results.append((col, 'categorical', float(chi2), float(p), np.nan, np.nan, f'degrees_of_freedom={dof}'))
    except Exception as e:
        results.append((col, 'categorical', np.nan, np.nan, np.nan, np.nan, f'error:{e}'))

# 7) Summarize results
res_df = pd.DataFrame(results, columns=[
    'feature','feature_type','test_statistic','p_value','mannwhitney_p','cohens_d_or_nan','notes'
])

# significance flag: numeric uses t-test p-value, categorical uses chi-square p-value
res_df['significant'] = False
res_df.loc[(res_df['feature_type']=='numeric') & (res_df['p_value'] < 0.05), 'significant'] = True
res_df.loc[(res_df['feature_type']=='categorical') & (res_df['p_value'] < 0.05), 'significant'] = True

# order numeric by significance then effect size
num_df = res_df[res_df['feature_type']=='numeric'].copy().sort_values(['significant','cohens_d_or_nan'], ascending=[False,False])
cat_df = res_df[res_df['feature_type']=='categorical'].copy().sort_values(['significant','p_value'], ascending=[False,True])
final_df = pd.concat([num_df, cat_df], ignore_index=True)

# 8) Save and display
out = Path("statistical_test_results.csv")
final_df.to_csv(out, index=False)
print("\nSaved results to:", out.resolve())
print("\nTop numeric features (by significance then Cohen's d):")
display(num_df.head(20))
print("\nTop categorical features (by significance):")
display(cat_df.head(20))

# 9) Plot top numeric significant features by |Cohen's d|
top_numeric = num_df[num_df['significant']].copy()
if not top_numeric.empty:
    top_numeric['abs_d'] = top_numeric['cohens_d_or_nan'].abs()
    top6 = top_numeric.sort_values('abs_d', ascending=False).head(6)['feature'].tolist()
    print("\nPlotting top numeric features:", top6)
    for col in top6:
        plt.figure(figsize=(6,4))
        data0 = group0[col].dropna()
        data1 = group1[col].dropna()
        plt.boxplot([data0, data1], labels=['Not Overstim (0)', 'Overstim (1)'])
        dval = float(num_df.loc[num_df['feature']==col, 'cohens_d_or_nan'].iloc[0])
        plt.title(f"{col} by Target (Cohen d = {dval:.3f})")
        plt.tight_layout()
        plt.savefig(Path("eda_plots") / f"stat_box_{col}.png", dpi=150)
        plt.show()
else:
    print("\nNo significant numeric features detected (p<0.05) to plot.")

print("\nStep 4 complete. Inspect 'statistical_test_results.csv' for the full table of results.")
```

```python
# ========== STEP 5: Class Imbalance Handling (SMOTE if available, else random oversample) ==========
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter
from pathlib import Path

# If you already have X and y defined, the code will use them.
# Otherwise it will derive X,y from df and target_col (if present).
if 'X' not in globals() or 'y' not in globals():
    if 'df' not in globals():
        raise FileNotFoundError("DataFrame 'df' not found. Run previous steps to create 'df'.")
    # robust target detection
    target_candidates = [c for c in df.columns if 'overstim' in c.lower()]
    if not target_candidates:
        raise KeyError("Target column containing 'overstim' not found in df.")
    target_col = target_candidates[0]
    X = df.drop(columns=[target_col])
    y = df[target_col]

print("Initial class distribution:", Counter(y))

# Ensure X is numeric: one-hot encode categorical features if present
X_enc = pd.get_dummies(X, drop_first=True)
print("Feature matrix shape after encoding:", X_enc.shape)

# Try to use SMOTE; if not available, fall back to random oversampling
use_smote = True
try:
    from imblearn.over_sampling import SMOTE
except Exception as e:
    print("SMOTE not available (or import failed). Will use random oversampling fallback. Error:", e)
    use_smote = False

if use_smote:
    print("Applying SMOTE to balance classes...")
    sm = SMOTE(random_state=42, k_neighbors=5)
    X_res, y_res = sm.fit_resample(X_enc, y)
else:
    # fallback: random oversample minority class to match majority size
    print("Applying random oversampling fallback (sample with replacement from minority class)...")
    y_series = pd.Series(y).reset_index(drop=True)
    X_enc = X_enc.reset_index(drop=True)
    counts = y_series.value_counts()
    majority_class = counts.idxmax()
    majority_count = counts.max()
    dfs = []
    for cls, cls_count in counts.items():
        X_cls = X_enc[y_series == cls].copy()
        y_cls = y_series[y_series == cls].copy()
        if cls_count < majority_count:
            # sample with replacement
            n_needed = majority_count - cls_count
            sampled = X_cls.sample(n=n_needed, replace=True, random_state=42)
            sampled_y = pd.Series([cls]*n_needed)
            X_new = pd.concat([X_cls, sampled], axis=0).reset_index(drop=True)
            y_new = pd.concat([y_cls.reset_index(drop=True), sampled_y], axis=0).reset_index(drop=True)
        else:
            X_new = X_cls.reset_index(drop=True)
            y_new = y_cls.reset_index(drop=True)
        dfs.append((X_new, y_new))
    X_res = pd.concat([d[0] for d in dfs], axis=0).reset_index(drop=True)
    y_res = pd.concat([d[1] for d in dfs], axis=0).reset_index(drop=True)

print("New class distribution:", Counter(y_res))

# Combine into balanced DataFrame
balanced_df = pd.concat([pd.DataFrame(X_res, columns=X_enc.columns).reset_index(drop=True),
                         pd.Series(y_res, name=(target_col if 'target_col' in globals() else 'Overstimulated')).reset_index(drop=True)],
                        axis=1)

print("Balanced dataset shape:", balanced_df.shape)

# Visualize before/after (requires original y)
plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
vc = Counter(y)
plt.bar([str(k) for k in vc.keys()], vc.values(), color=['teal','orange'])
plt.title("Before balancing")
for i,v in enumerate(vc.values()):
    plt.text(i, v + max(vc.values())*0.02, str(v), ha='center')
plt.subplot(1,2,2)
vc2 = Counter(y_res)
plt.bar([str(k) for k in vc2.keys()], vc2.values(), color=['teal','orange'])
plt.title("After balancing")
for i,v in enumerate(vc2.values()):
    plt.text(i, v + max(vc2.values())*0.02, str(v), ha='center')
plt.tight_layout()
plt.show()

# Save balanced dataset
out = Path("balanced_overstimulation_dataset.csv")
balanced_df.to_csv(out, index=False)
print("Saved balanced dataset to:", out.resolve())

# Put X_res, y_res, df_balanced into globals for later steps
X_res, y_res, df_balanced = X_res, y_res, balanced_df
```

```python
# ============================================
#  STEP 8 & 9: Train/Test Split, CV, Model Building & Evaluation
# ============================================

import os
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    classification_report, confusion_matrix, roc_auc_score, roc_curve,
    accuracy_score, precision_score, recall_score, f1_score
)
import joblib

# ---------- Choose dataset (preference order) ----------
# Uses data prepared in previous steps if present.
if 'X_std_scaled' in globals() and 'y_scaled' in globals():
    X = X_std_scaled.copy()
    y = y_scaled.copy()
    print("Using Standard-scaled features (X_std_scaled).")
elif 'X_mm_scaled' in globals() and 'y_scaled' in globals():
    X = X_mm_scaled.copy()
    y = y_scaled.copy()
    print("Using MinMax-scaled features (X_mm_scaled).")
elif 'X_selected' in globals() and 'y_selected' in globals():
    X = X_selected.copy()
    y = y_selected.copy()
    print("Using X_selected / y_selected.")
elif 'X_all_features' in globals() and 'y_all_features' in globals():
    X = X_all_features.copy()
    y = y_all_features.copy()
    print("Using X_all_features / y_all_features.")
elif 'df_balanced' in globals():
    # fallback: create X,y from df_balanced
    tgt = [c for c in df_balanced.columns if 'overstim' in c.lower()][0]
    X = df_balanced.drop(columns=[tgt]).select_dtypes(include=[np.number])
    y = df_balanced[tgt]
    print("Using numeric columns from df_balanced.")
else:
    raise RuntimeError("No suitable X/y found in the notebook environment. Ensure you ran Step 7/6 outputs.")

# Convert to numpy if necessary
if isinstance(X, pd.DataFrame):
    feature_names = X.columns.tolist()
else:
    feature_names = [f"f{i}" for i in range(X.shape[1])]
    X = pd.DataFrame(X, columns=feature_names)

y = pd.Series(y).astype(int).reset_index(drop=True)
X = X.reset_index(drop=True)

print("Feature matrix shape:", X.shape)
print("Target distribution:", y.value_counts().to_dict())

# ---------- 1) Train-test split (stratified) ----------
test_size = 0.2
RANDOM_STATE = 42
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=test_size, stratify=y, random_state=RANDOM_STATE
)
print(f"\nTrain shape: {X_train.shape}, Test shape: {X_test.shape}")

# ---------- 2) Stratified K-Fold Cross-Validation ----------
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

models = {
    'LogisticRegression': LogisticRegression(max_iter=2000, solver='liblinear', random_state=RANDOM_STATE),
    'RandomForest': RandomForestClassifier(n_estimators=200, random_state=RANDOM_STATE, n_jobs=-1)
}

cv_results = {}
for name, model in models.items():
    print(f"\nCross-validating {name} ...")
    # F1-score (average binary)
    f1_scores = cross_val_score(model, X_train, y_train, cv=cv, scoring='f1', n_jobs=-1)
    # ROC AUC
    roc_scores = cross_val_score(model, X_train, y_train, cv=cv, scoring='roc_auc', n_jobs=-1)
    cv_results[name] = {
        'f1_mean': float(np.mean(f1_scores)), 'f1_std': float(np.std(f1_scores)),
        'roc_auc_mean': float(np.mean(roc_scores)), 'roc_auc_std': float(np.std(roc_scores))
    }
    print(f"  F1: {cv_results[name]['f1_mean']:.4f} ± {cv_results[name]['f1_std']:.4f}")
    print(f"  ROC-AUC: {cv_results[name]['roc_auc_mean']:.4f} ± {cv_results[name]['roc_auc_std']:.4f}")

# ---------- 3) Train models on full training set and evaluate on test set ----------
results = {}
out_dir = Path("models_and_results")
out_dir.mkdir(exist_ok=True)

for name, model in models.items():
    print(f"\nTraining {name} on full training set...")
    model.fit(X_train, y_train)
    # Save trained model
    model_path = out_dir / f"{name}.pkl"
    joblib.dump(model, model_path)
    print(f"Saved model to {model_path}")

    # Predictions
    y_pred = model.predict(X_test)
    if hasattr(model, "predict_proba"):
        y_proba = model.predict_proba(X_test)[:, 1]
    else:
        # some models provide decision_function
        try:
            y_proba = model.decision_function(X_test)
            # if shape is 1D, convert to probabilities via sigmoid
            if y_proba.ndim == 1:
                y_proba = 1 / (1 + np.exp(-y_proba))
        except Exception:
            y_proba = np.zeros_like(y_pred)

    # Metrics
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, zero_division=0)
    rec = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)
    roc_auc = roc_auc_score(y_test, y_proba)

    # Confusion matrix
    cm = confusion_matrix(y_test, y_pred)

    # Save metrics & reports
    results[name] = {
        'accuracy': float(acc),
        'precision': float(prec),
        'recall': float(rec),
        'f1': float(f1),
        'roc_auc': float(roc_auc),
        'confusion_matrix': cm,
        'classification_report': classification_report(y_test, y_pred, digits=4, output_dict=True),
        'cv_results': cv_results.get(name, {})
    }

    # Print summary
    print(f"\n{name} Test results:")
    print(f"  Accuracy:  {acc:.4f}")
    print(f"  Precision: {prec:.4f}")
    print(f"  Recall:    {rec:.4f}")
    print(f"  F1:        {f1:.4f}")
    print(f"  ROC-AUC:   {roc_auc:.4f}")
    print("\nClassification report:\n", classification_report(y_test, y_pred, digits=4))
    print("Confusion matrix:\n", cm)

    # ---------- Plots ----------
    # Confusion matrix heatmap
    plt.figure(figsize=(5,4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
                xticklabels=['Not Overstim (0)','Overstim (1)'],
                yticklabels=['Not Overstim (0)','Overstim (1)'])
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.title(f"{name} — Confusion Matrix")
    plt.tight_layout()
    plt.savefig(out_dir / f"{name}_confusion_matrix.png", dpi=150)
    plt.show()

    # ROC curve
    fpr, tpr, _ = roc_curve(y_test, y_proba)
    plt.figure(figsize=(6,5))
    plt.plot(fpr, tpr, label=f"{name} (AUC = {roc_auc:.3f})")
    plt.plot([0,1],[0,1],'k--', alpha=0.4)
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title(f"{name} — ROC Curve")
    plt.legend(loc='lower right')
    plt.tight_layout()
    plt.savefig(out_dir / f"{name}_roc_curve.png", dpi=150)
    plt.show()

# ---------- Save aggregated metrics to CSV ----------
metrics_rows = []
for name, r in results.items():
    cr = r['classification_report']
    metrics_rows.append({
        'model': name,
        'accuracy': r['accuracy'],
        'precision': r['precision'],
        'recall': r['recall'],
        'f1': r['f1'],
        'roc_auc': r['roc_auc'],
        'cv_f1_mean': r['cv_results'].get('f1_mean', np.nan),
        'cv_f1_std': r['cv_results'].get('f1_std', np.nan),
        'cv_roc_mean': r['cv_results'].get('roc_auc_mean', np.nan),
        'cv_roc_std': r['cv_results'].get('roc_auc_std', np.nan)
    })

metrics_df = pd.DataFrame(metrics_rows)
metrics_df.to_csv(out_dir / "model_metrics_summary.csv", index=False)
print("\nSaved model metrics summary to:", out_dir / "model_metrics_summary.csv")

# ---------- Save predictions (optional) ----------
preds_df = pd.DataFrame({
    'y_test': y_test.reset_index(drop=True),
    'LogReg_pred': joblib.load(out_dir / "LogisticRegression.pkl").predict(X_test),
    'RF_pred': joblib.load(out_dir / "RandomForest.pkl").predict(X_test)
})
preds_df.to_csv(out_dir / "test_set_predictions.csv", index=False)
print("Saved test set predictions to:", out_dir / "test_set_predictions.csv")

print("\nAll models trained & evaluated. Artifacts saved in:", out_dir.resolve())
```

```python
# ============================================
#  STEP 10 (updated) & STEP 11 (tuning): Compare models and tune RandomForest
# ============================================

import warnings
warnings.filterwarnings("ignore")
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score, roc_auc_score,
    confusion_matrix, classification_report, roc_curve
)
from scipy.stats import randint, uniform

# -----------------------------
out_dir = Path("models_and_results")
out_dir.mkdir(exist_ok=True)

# 1) Load previous metrics if available
metrics_path = out_dir / "model_metrics_summary.csv"
if metrics_path.exists():
    metrics_df = pd.read_csv(metrics_path)
else:
    # Build metrics_df from in-memory `results` if available
    if 'results' in globals():
        rows = []
        for name, r in results.items():
            rows.append({'model': name,
                         'accuracy': r['accuracy'],
                         'precision': r['precision'],
                         'recall': r['recall'],
                         'f1': r['f1'],
                         'roc_auc': r['roc_auc']})
        metrics_df = pd.DataFrame(rows)
        metrics_df.to_csv(metrics_path, index=False)
    else:
        raise FileNotFoundError("No metrics file or in-memory results found. Run Step 8/9 first.")
print("Loaded metrics:\n")
display(metrics_df)

# 2) Visual comparison (accuracy/precision/recall/f1/roc_auc)
melted = metrics_df.melt(id_vars='model', value_vars=['accuracy','precision','recall','f1','roc_auc'],
                         var_name='metric', value_name='score')
plt.figure(figsize=(10,5))
sns.barplot(data=melted, x='metric', y='score', hue='model')
plt.ylim(0,1)
plt.title("Model Comparison: Key Metrics")
plt.tight_layout()
plt.savefig(out_dir / "model_comparison_metrics.png", dpi=150)
plt.show()

# 3) Select best model by roc_auc
best_row = metrics_df.sort_values('roc_auc', ascending=False).iloc[0]
best_model_name = best_row['model']
print("Best model by ROC-AUC:", best_model_name)

# 4) Ensure training/test data available
if 'X_train' not in globals() or 'X_test' not in globals() or 'y_train' not in globals() or 'y_test' not in globals():
    # try to reconstruct from earlier variables
    if 'X' in globals() and 'y' in globals():
        from sklearn.model_selection import train_test_split
        RANDOM_STATE = 42
        X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2,
                                                            stratify=y, random_state=RANDOM_STATE)
    elif 'df' in globals():
        tgt = [c for c in df.columns if 'overstim' in c.lower()][0]
        X_all = df.drop(columns=[tgt]).select_dtypes(include=[np.number])
        y_all = df[tgt]
        from sklearn.model_selection import train_test_split
        RANDOM_STATE = 42
        X_train, X_test, y_train, y_test = train_test_split(X_all, y_all, test_size=0.2,
                                                            stratify=y_all, random_state=RANDOM_STATE)
    else:
        raise RuntimeError("Train/test data not found. Please re-run Step 8 to produce X_train/X_test/y_train/y_test.")

# 5) Load baseline RandomForest (if saved), else create new baseline and evaluate
rf_path = out_dir / "RandomForest.pkl"
if rf_path.exists():
    baseline_rf = joblib.load(rf_path)
    print("Loaded saved RandomForest baseline from:", rf_path)
else:
    baseline_rf = RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1)
    baseline_rf.fit(X_train, y_train)
    joblib.dump(baseline_rf, rf_path)
    print("Trained & saved baseline RandomForest to:", rf_path)

def evaluate_model(name, model, X_test, y_test, save_prefix):
    y_pred = model.predict(X_test)
    try:
        y_proba = model.predict_proba(X_test)[:,1]
    except Exception:
        y_proba = model.decision_function(X_test)
        if y_proba.ndim == 1:
            y_proba = 1/(1+np.exp(-y_proba))
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, zero_division=0)
    rec = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)
    roc = roc_auc_score(y_test, y_proba)
    cm = confusion_matrix(y_test, y_pred)
    cr = classification_report(y_test, y_pred, digits=4)
    # save/confusion plot
    plt.figure(figsize=(5,4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
                xticklabels=['Not Overstim (0)','Overstim (1)'],
                yticklabels=['Not Overstim (0)','Overstim (1)'])
    plt.title(f"{name} — Confusion Matrix")
    plt.xlabel("Predicted"); plt.ylabel("Actual")
    plt.tight_layout()
    plt.savefig(out_dir / f"{save_prefix}_confusion_matrix.png", dpi=150)
    plt.show()
    # ROC
    fpr, tpr, _ = roc_curve(y_test, y_proba)
    plt.figure(figsize=(6,5))
    plt.plot(fpr, tpr, label=f"{name} (AUC = {roc:.3f})")
    plt.plot([0,1],[0,1],'k--', alpha=0.4)
    plt.xlabel("False Positive Rate"); plt.ylabel("True Positive Rate")
    plt.title(f"{name} — ROC Curve")
    plt.legend(loc='lower right')
    plt.tight_layout()
    plt.savefig(out_dir / f"{save_prefix}_roc_curve.png", dpi=150)
    plt.show()
    return {'accuracy':acc,'precision':prec,'recall':rec,'f1':f1,'roc_auc':roc,'confusion_matrix':cm,'classification_report':cr}

print("\nBaseline RandomForest evaluation on test set:")
baseline_results = evaluate_model("RandomForest_baseline", baseline_rf, X_test, y_test, "RF_baseline")
print(pd.Series({k: baseline_results[k] for k in ['accuracy','precision','recall','f1','roc_auc']}))

# 6) Hyperparameter tuning using RandomizedSearchCV for RandomForest (because it was best)
print("\nStarting RandomizedSearchCV tuning for RandomForest...")

param_dist = {
    "n_estimators": randint(100, 500),
    "max_depth": [None] + list(range(6, 31, 6)),
    "min_samples_split": randint(2, 11),
    "min_samples_leaf": randint(1, 7),
    "max_features": ["sqrt", "log2", None],
    "bootstrap": [True, False]
}

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
rs = RandomizedSearchCV(
    estimator=RandomForestClassifier(random_state=42, n_jobs=-1),
    param_distributions=param_dist,
    n_iter=40,
    scoring='roc_auc',
    cv=cv,
    verbose=2,
    random_state=42,
    n_jobs=-1
)
rs.fit(X_train, y_train)

print("\nBest parameters from RandomizedSearchCV:")
print(rs.best_params_)
print("Best CV ROC-AUC:", rs.best_score_)

# 7) Retrain best estimator on full training set
best_rf = rs.best_estimator_
best_rf.fit(X_train, y_train)
joblib.dump(best_rf, out_dir / "RandomForest_tuned.pkl")
print("Saved tuned RandomForest to:", out_dir / "RandomForest_tuned.pkl")

# 8) Evaluate tuned model on test set and compare to baseline
print("\nTuned RandomForest evaluation on test set:")
tuned_results = evaluate_model("RandomForest_tuned", best_rf, X_test, y_test, "RF_tuned")
print("\nBaseline metrics:", {k: round(v,4) for k,v in baseline_results.items() if k in ['accuracy','f1','roc_auc']})
print("Tuned metrics:   ", {k: round(v,4) for k,v in tuned_results.items() if k in ['accuracy','f1','roc_auc']})

# 9) Save comparison table
comparison = pd.DataFrame([
    {"model":"RandomForest_baseline", **{k: baseline_results[k] for k in ['accuracy','precision','recall','f1','roc_auc']}},
    {"model":"RandomForest_tuned",    **{k: tuned_results[k] for k in ['accuracy','precision','recall','f1','roc_auc']}}
])
comparison.to_csv(out_dir / "rf_baseline_vs_tuned_comparison.csv", index=False)
print("\nSaved comparison table to:", out_dir / "rf_baseline_vs_tuned_comparison.csv")
display(comparison)

# 10) If tuned improved, promote it: update model_metrics_summary.csv
try:
    # update or append
    metrics_df = pd.read_csv(metrics_path)
    # Remove older RandomForest row(s)
    metrics_df = metrics_df[metrics_df['model'] != 'RandomForest']
    # Append tuned RF results
    tuned_row = {
        'model': 'RandomForest_tuned',
        'accuracy': float(tuned_results['accuracy']),
        'precision': float(tuned_results['precision']),
        'recall': float(tuned_results['recall']),
        'f1': float(tuned_results['f1']),
        'roc_auc': float(tuned_results['roc_auc'])
    }
    metrics_df = pd.concat([metrics_df, pd.DataFrame([tuned_row])], ignore_index=True)
    metrics_df.to_csv(metrics_path, index=False)
    print("Updated metrics file with tuned RandomForest.")
except Exception as e:
    print("Could not update metrics file:", e)

print("\nDone — tuned model saved and comparison exported to models_and_results/.")
```
